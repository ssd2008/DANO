```python
target = 'target'

def categorial_influance(df, cat, target=target):
    return pd.pivot_table(df, index=cat, columns=target, values=target, aggfunc='count')

def numeric_influance(df, num, target=target):
    return pd.pivot_table(df, index=target, values=num, aggfunc='mean')
```
